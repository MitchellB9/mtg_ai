```python
# imports
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np

# Make pandas usable for wide MTG data
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", 120)
pd.set_option("display.width", 0)
```

```python
# confirm path to data
PROJECT_ROOT = Path("../../..").resolve()
print(PROJECT_ROOT)
DATA_DIR = PROJECT_ROOT / "data" / "base"
LINKS_IDS_PATH = DATA_DIR / "links_and_ids.parquet"
print(LINKS_IDS_PATH)

# Load parquet to dataframe
cards_links  = pd.read_parquet(LINKS_IDS_PATH)
# Basic shape and schema
print(cards_links .shape)
cards_links .head(5)
```

```python
cards_links.info()
```

```python
# Identity checks
print(f"Rows: {len(cards_links):,}")
print(f"Unique id: {cards_links['id'].nunique():,}")
print(f"Unique oracle_id: {cards_links['oracle_id'].nunique():,}")

print(f"Duplicate id rows: {cards_links['id'].duplicated().sum():,}")
print(f"Duplicate oracle_id rows: {cards_links['oracle_id'].duplicated().sum():,}")
```

```python
# Missing values overview
missing = cards_links.isnull().sum().sort_values(ascending=False)

missing[missing > 0]
```

```python
# Coverage of major external IDs
id_cols = [
    "multiverse_ids",
    "mtgo_id",
    "arena_id",
    "tcgplayer_id",
    "cardmarket_id",
]

coverage = {
    col: cards_links[col].notna().sum()
    for col in id_cols
    if col in cards_links.columns
}

coverage
```

```python
# URI coverage
uri_cols = [
    "uri",
    "scryfall_uri",
    "set_uri",
    "set_search_uri",
    "scryfall_set_uri",
    "rulings_uri",
    "prints_search_uri",
]

uri_coverage = {
    col: cards_links[col].notna().sum()
    for col in uri_cols
    if col in cards_links.columns
}

uri_coverage
```

```python
# Image URI inspection
if "image_uris" in cards_links.columns:
    image_rows = cards_links["image_uris"].notna().sum()

    print(f"Rows with image_uris: {image_rows:,}")

    display(
        cards_links.loc[
            cards_links["image_uris"].notna(),
            ["name", "image_uris"]
        ].head(5)
    )
```
