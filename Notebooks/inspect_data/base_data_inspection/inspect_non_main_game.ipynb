```python
# imports
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np

# Make pandas usable for wide MTG data
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", 120)
pd.set_option("display.width", 0)
```

```python
# confirm path to data
PROJECT_ROOT = Path("../../..").resolve()
print(PROJECT_ROOT)
DATA_DIR = PROJECT_ROOT / "data" / "base"
NON_MAIN_PATH = DATA_DIR / "non_main_game_pieces.parquet"
print(NON_MAIN_PATH)

# Load parquet to dataframe
cards_non_main = pd.read_parquet(NON_MAIN_PATH)
# Basic shape and schema
print(cards_non_main.shape)
cards_non_main.head(5)
```

```python
cards_non_main.info()
```

```python
# Identity checks
print(f"Rows: {len(cards_non_main):,}")

for col in ["id", "oracle_id", "name"]:
    if col in cards_non_main.columns:
        print(f"Unique {col}: {cards_non_main[col].nunique():,}")
        print(f"Duplicate {col} rows: {cards_non_main[col].duplicated().sum():,}")
```

```python
# Missing values overview
missing = cards_non_main.isnull().sum().sort_values(ascending=False)
missing[missing > 0]
```

```python
# Layout distribution
if "layout" in cards_non_main.columns:
    layout_counts = cards_non_main["layout"].value_counts()

    display(layout_counts)

    layout_counts.plot(kind="bar", figsize=(12, 5))
    plt.title("Non-Main-Game Layout Distribution")
    plt.ylabel("Count")
    plt.xticks(rotation=45)
    plt.show()
```

```python
# Type line distribution
if "type_line" in cards_non_main.columns:
    type_line_counts = cards_non_main["type_line"].value_counts().head(40)

    display(type_line_counts)

    type_line_counts.plot(kind="bar", figsize=(14, 5))
    plt.title("Top Non-Main-Game Type Lines")
    plt.ylabel("Count")
    plt.xticks(rotation=75)
    plt.show()
```

```python
# Set type distribution
if "set_type" in cards_non_main.columns:
    set_type_counts = cards_non_main["set_type"].value_counts()

    display(set_type_counts)

    set_type_counts.plot(kind="bar", figsize=(12, 5))
    plt.title("Non-Main-Game Set Type Distribution")
    plt.ylabel("Count")
    plt.xticks(rotation=45)
    plt.show()
```

```python
# Border color distribution
if "border_color" in cards_non_main.columns:
    border_counts = cards_non_main["border_color"].value_counts()

    display(border_counts)

    border_counts.plot(kind="bar", figsize=(8, 4))
    plt.title("Non-Main-Game Border Color Distribution")
    plt.ylabel("Count")
    plt.show()
```

```python
# Security stamp / acorn check
if "security_stamp" in cards_non_main.columns:
    security_counts = cards_non_main["security_stamp"].value_counts(dropna=False)

    display(security_counts)

    security_counts.plot(kind="bar", figsize=(8, 4))
    plt.title("Security Stamp Distribution")
    plt.ylabel("Count")
    plt.xticks(rotation=45)
    plt.show()
```

```python
# Explicit Token/Card type-line check
if "type_line" in cards_non_main.columns:
    token_card_mask = cards_non_main["type_line"].fillna("").str.contains(
        r"\b(?:Token|Card)\b",
        case=False,
        regex=True,
    )

    print(f"Rows with Token/Card in type_line: {token_card_mask.sum():,}")

    display(
        cards_non_main.loc[
            token_card_mask,
            ["id", "oracle_id", "name", "layout", "type_line", "set", "set_name"]
        ].head(50)
    )
```

```python
# Look for non-main-game indicators by layout/type/set
inspection_cols = [
    "id", "oracle_id", "name", "layout", "type_line",
    "set", "set_name", "set_type", "border_color", "security_stamp"
]
inspection_cols = [col for col in inspection_cols if col in cards_non_main.columns]

cards_non_main[inspection_cols].sample(25, random_state=42)
```

```python
# Rows that may have been sequestered only because of layout
if "layout" in cards_non_main.columns and "type_line" in cards_non_main.columns:
    layout_only_mask = (
        cards_non_main["layout"].notna()
        & ~cards_non_main["type_line"].fillna("").str.contains(
            r"\b(?:Token|Card)\b",
            case=False,
            regex=True,
        )
    )

    display(cards_non_main.loc[layout_only_mask, inspection_cols].head(50))
```

```python
# Possible future subgroups inside non-main-game pieces
if "layout" in cards_non_main.columns:
    for layout in cards_non_main["layout"].value_counts().index:
        print(f"\n--- Layout: {layout} ---")
        display(cards_non_main.loc[cards_non_main["layout"] == layout, inspection_cols].head(5))
```
