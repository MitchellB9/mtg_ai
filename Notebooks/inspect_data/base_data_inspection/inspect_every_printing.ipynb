```python
# imports
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np

# Make pandas usable for wide MTG data
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", 120)
pd.set_option("display.width", 0)
```

```python
# confirm path to data
PROJECT_ROOT = Path("../../..").resolve()
print(PROJECT_ROOT)
DATA_DIR = PROJECT_ROOT / "data" / "base"
CARDS_EVERY_PATH = DATA_DIR / "every_printing.parquet"
print(CARDS_EVERY_PATH)

# Load parquet to dataframe
cards_every = pd.read_parquet(CARDS_EVERY_PATH)
# Basic shape and schema
print(cards_every.shape)
cards_every.head(5)
```

```python
cards_every.info()
```

```python
# Identity checks
print(f"Rows: {len(cards_every):,}")
print(f"Unique id: {cards_every['id'].nunique():,}")
print(f"Unique oracle_id: {cards_every['oracle_id'].nunique():,}")

print(f"Duplicate id rows: {cards_every['id'].duplicated().sum():,}")
print(f"Duplicate oracle_id rows: {cards_every['oracle_id'].duplicated().sum():,}")
```

```python
# Missing values overview
missing = cards_every.isnull().sum().sort_values(ascending=False)

missing[missing > 0]
```

```python
# Layout distribution
layout_counts = cards_every["layout"].value_counts()

display(layout_counts)

fig, (ax_top, ax_bottom) = plt.subplots(
    2,
    1,
    sharex=True,
    figsize=(12, 7),
    gridspec_kw={"height_ratios": [1, 3]}
)

layout_counts.plot(kind="bar", ax=ax_top)
layout_counts.plot(kind="bar", ax=ax_bottom)

non_normal_counts = layout_counts.drop(labels=["normal"], errors="ignore")
max_non_normal = non_normal_counts.max()

ax_top.set_ylim(layout_counts.max() * 0.90, layout_counts.max() * 1.05)
ax_bottom.set_ylim(0, max_non_normal * 1.15)

ax_top.spines["bottom"].set_visible(False)
ax_bottom.spines["top"].set_visible(False)

ax_bottom.tick_params(axis="x", rotation=45)

ax_top.set_title("Printing Layout Distribution")
ax_bottom.set_ylabel("Count")

plt.tight_layout()
plt.show()
```

```python
# Set type distribution
set_type_counts = cards_every["set_type"].value_counts()

display(set_type_counts)

set_type_counts.plot(kind="bar", figsize=(12, 5))
plt.title("Set Type Distribution")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
# Promo vs non-promo
if "promo" in cards_every.columns:
    promo_counts = cards_every["promo"].value_counts()

    display(promo_counts)

    promo_counts.plot(kind="bar", figsize=(5, 4))
    plt.title("Promo Distribution")
    plt.ylabel("Count")
    plt.show()
```

```python
# Reprint distribution
if "reprint" in cards_every.columns:
    reprint_counts = cards_every["reprint"].value_counts()

    display(reprint_counts)

    reprint_counts.plot(kind="bar", figsize=(5, 4))
    plt.title("Reprint Distribution")
    plt.ylabel("Count")
    plt.show()
```

```python
# Language distribution
lang_counts = cards_every["lang"].value_counts().head(20)

display(lang_counts)

lang_counts.plot(kind="bar", figsize=(10, 4))
plt.title("Top Languages")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
# Most printed cards
most_printed = (
    cards_every.groupby(["name"])
    .size()
    .reset_index(name="printing_count")
    .sort_values("printing_count", ascending=False)
)

display(most_printed.head(25))
```

```python
# Release year distribution
cards_every["release_year"] = pd.to_datetime(
    cards_every["released_at"],
    errors="coerce"
).dt.year

year_counts = cards_every["release_year"].value_counts().sort_index()

year_counts.plot(figsize=(12, 5))
plt.title("Printings by Release Year")
plt.ylabel("Count")
plt.show()
```
